# 1. Notebooks vs Python source files

## What a notebook actually is

The file you are reading is `ecosystem.ipynb`. Despite appearances it is **not** a
document; it is a JSON file listing cells, and a separate program called the **kernel**
(a Python process running in the background) executes the code cells one at a time and
sends the results back to be displayed.

Three consequences follow from that, and all three bite beginners:

**1. The kernel remembers everything.** Variables live in the kernel, not in the file. If
you define `x = 5` and then delete the cell, `x` still exists. The only way to truly clear
it is **Kernel → Restart**.

**2. Execution order is whatever you clicked, not top to bottom.** The `[1]`, `[2]`,
`[3]` numbers to the left of each cell show the order you actually ran them. A notebook
where those numbers are out of order is a notebook whose results you cannot trust.

**3. A notebook that "works" may not work for anybody else.** The classic disaster: you
define a variable, edit the cell to something else, and never re-run it. Your kernel still
holds the old value; a fresh kernel will crash.

> **The rule that saves you:** before you share, submit, or trust a notebook, run
> **Kernel → Restart Kernel and Run All Cells**. If it survives that, it is real.

## Notebook superpowers

**Magics** are special commands starting with `%` (one line) or `%%` (whole cell):

| Magic | What it does |
|---|---|
| `%whos` | list every variable currently defined |
| `%time expr` / `%timeit expr` | time a statement (`%timeit` repeats for accuracy) |
| `%%writefile name.py` | save the rest of the cell to a file |
| `%run script.py` | run a script inside this kernel |
| `%pip install pkg` | install into *this kernel's* environment (see section 2) |
| `%reset -f` | wipe all variables without restarting |

A leading `!` runs a **shell** command: `!python --version`, `!ls`, `!dir`.

Other niceties: `?` after any name shows its documentation (`len?`), **Tab** completes
names, and **Shift + Tab** inside a call shows the signature.

## What notebooks are bad at

- **Version control.** A notebook diff is a wall of JSON. Git cannot review it usefully.
- **Reuse.** You cannot `import` a function from a notebook into another project.
- **Testing and automation.** No test runner or scheduler wants an `.ipynb`.
- **Long programs.** Anything over a few hundred lines becomes unnavigable.

## Python source files

A `.py` file is just text containing Python code. Run it from a terminal:

```
python analysis.py
```

It runs top to bottom, then exits. Nothing is displayed unless you `print` it, and every
variable disappears when it finishes.

The real payoff is that a `.py` file is a **module**: any other file can import it.

```python
# in mymath.py
def area_of_circle(r):
    return 3.141592653589793 * r ** 2

# anywhere else; note: no .py in the import
import mymath
mymath.area_of_circle(2)

from mymath import area_of_circle
area_of_circle(2)
```

## The `if __name__ == "__main__":` guard

**Importing a module runs every top-level line in it.** So if `mymath.py` ends with
`print(area_of_circle(5))`, that line fires the moment somebody imports it; which is
almost never what you want.

The guard fixes this. Python sets the variable `__name__` to `"__main__"` in the file you
ran directly, and to the module's name in anything that was imported. So:

```python
def area_of_circle(r):
    return 3.141592653589793 * r ** 2

if __name__ == "__main__":
    # runs on `python mymath.py`, skipped on `import mymath`
    print(area_of_circle(5))
```

Put definitions at the top, the script's actual *actions* inside the guard. This is one of
the strongest conventions in Python; follow it in every file you write.

## So which do I use?

| Use a notebook for | Use a `.py` file for |
|---|---|
| exploring data for the first time | functions you will reuse |
| anything with plots | anything anyone else will run |
| teaching, reports, showing your working | tested, automated or scheduled code |
| quick throwaway calculations | programs longer than a screen or two |

In practice, professionals use both: **explore in a notebook, then promote the code that
turned out to be useful into a `.py` module** and import it back. That workflow gives you
the interactivity *and* the reusability.

# 2. Environments

## The problem

You finish a project that uses `numpy` version 1.24. Six months later you start a new one
that needs `numpy` 2.1. You upgrade, and the old project breaks. Now scale that to a
dozen projects, each with twenty packages, each with their own dependencies, and you have
the reason virtual environments exist.

The other half of the problem is sharing. "It works on my machine" almost always means
*my machine has packages yours does not, and I have no idea which ones*.

## What an environment actually is

A **virtual environment** is a folder containing its own copy (or link) of the Python
interpreter plus its own `site-packages` directory where installed libraries go.
"Activating" it just puts that folder first on your `PATH`, so `python` and `pip` mean
*those* ones.

That is the whole trick. One project, one folder, one isolated set of packages. Delete the
folder and the environment is gone, with no trace left on the rest of your system.

## `pip` and PyPI

**PyPI** (the Python Package Index, <https://pypi.org>) is the public repository of
several hundred thousand packages. **`pip`** is the tool that downloads and installs from
it.

```
pip install numpy               # latest version
pip install "numpy==1.26.4"     # an exact version
pip install -U numpy            # upgrade
pip uninstall numpy
pip list                        # everything installed here
pip show numpy                  # version, location, dependencies
```

## Creating and using a venv

`venv` ships with Python; nothing to install.

```
python -m venv .venv                  # create it (a folder called .venv)
```

Then activate it:

| Shell | Command |
|---|---|
| Windows PowerShell | `.venv\Scripts\Activate.ps1` |
| Windows cmd | `.venv\Scripts\activate.bat` |
| macOS / Linux | `source .venv/bin/activate` |

Your prompt gains a `(.venv)` prefix. From then on `pip install` puts packages inside that
folder. `deactivate` leaves.

Add `.venv/` to your `.gitignore`; you never commit an environment, you commit the
*recipe* for one.

## Reproducibility: `requirements.txt`

```
pip freeze > requirements.txt        # record exactly what is installed
pip install -r requirements.txt      # recreate it elsewhere
```

That file is what you commit, share, and hand to a marker. Pin versions (`numpy==2.1.0`)
when the result has to be reproducible; which, in scientific work, is always.

## Other tools you will hear about

- **conda / miniconda / mamba**: an alternative that manages *non-Python* dependencies
  too (compilers, CUDA, MKL). Very common in scientific computing, because installing
  something like PyTorch with GPU support used to be miserable with pip alone.
  Its equivalents: `conda create -n myenv python=3.11`, `conda activate myenv`.
- **uv**: a much faster modern replacement for pip/venv, gaining ground quickly.
- **poetry / pdm / pyproject.toml**: dependency management for packages you intend to
  publish.

They all solve the same underlying problem. Learn `venv` + `pip` first; the concepts
transfer directly.

## The notebook trap you must know about

A Jupyter kernel is tied to **one specific Python installation**. The `!pip install`
shell command may run a *different* pip, belonging to a different Python, and the package
lands somewhere your kernel cannot see. The symptom is maddening: pip says
"Successfully installed", and `import` still fails.

**Always use the `%pip` magic instead**; it is guaranteed to install into the kernel's
own environment:

```python
%pip install pandas        # correct
!pip install pandas        # roulette
```

To see which Python your kernel actually is, print `sys.executable`.

import math                     # use as math.sqrt(2)

## 3a. numpy: numerical arrays

### Why it exists

A Python list can hold anything, so every element carries type information and a pointer.
A **numpy array** holds one type, packed in a contiguous block of memory, and its
operations run in compiled C. For numerical work that is typically **10 to 100 times
faster**, with less memory and much shorter code.

```python
import numpy as np
```

### Creating arrays

```python
np.array([1, 2, 3])                # from a list
np.array([[1, 2], [3, 4]])         # 2D, from nested lists
np.zeros(5)                        # [0. 0. 0. 0. 0.]
np.ones((2, 3))                    # 2x3 of ones
np.full((2, 2), 7)
np.eye(3)                          # 3x3 identity
np.arange(0, 10, 2)                # like range: [0 2 4 6 8]
np.linspace(0, 1, 5)               # 5 evenly spaced points INCLUDING both ends
np.random.default_rng(0).normal(size=10)    # random samples
```

`arange` takes a *step*; `linspace` takes a *count*. For plotting an x-axis you almost
always want `linspace`.

### Inspecting an array

`a.shape` (dimensions), `a.ndim` (how many), `a.size` (total elements), `a.dtype` (the
element type, e.g. `float64` or `int64`).

### Vectorised arithmetic: the whole point

Operations apply **element-wise, with no loop**:

```python
a = np.array([1, 2, 3, 4])
a * 2            # [2 4 6 8]
a + a            # [2 4 6 8]
a ** 2           # [1 4 9 16]
np.sqrt(a)       # element-wise sqrt
np.sin(a)        # element-wise sin
```

> **The numpy mindset:** if you are writing a `for` loop over an array, there is almost
> always a vectorised expression that is shorter, clearer and far faster. Reach for the
> loop only when you truly cannot avoid it.

### Indexing, slicing and masks

```python
a[0], a[-1], a[1:3]                 # like lists
m = np.array([[1, 2, 3], [4, 5, 6]])
m[1, 2]                             # row 1, column 2 -> 6
m[0]                                # first row
m[:, 1]                             # ALL rows, column 1 -> [2 5]

a[a > 2]                            # boolean mask -> only elements greater than 2
(a > 2).sum()                       # how many satisfy it (True counts as 1)
np.where(a > 2, a, 0)               # element-wise if/else
```

Boolean masking is the single most useful numpy idiom. Learn it properly.

### Aggregation and `axis`

```python
a.sum(), a.mean(), a.std(), a.min(), a.max()
a.argmax()               # the INDEX of the largest element

m.sum()                  # everything
m.sum(axis=0)            # collapse the rows -> one total per COLUMN
m.sum(axis=1)            # collapse the columns -> one total per ROW
```

`axis` is the dimension being *eliminated*. That one sentence clears up most confusion.

### Reshaping and broadcasting

```python
np.arange(6).reshape(2, 3)       # (-1 means "work it out": .reshape(-1, 3))
m.T                              # transpose
m @ m.T                          # matrix multiplication

np.array([[1], [2], [3]]) + np.array([10, 20])   # shapes (3,1) and (2,) -> (3,2)
```

**Broadcasting** is numpy stretching smaller arrays across bigger ones so their shapes
line up. `a * 2` is the simplest case. It is powerful and occasionally baffling; when a
shape error appears, print `.shape` on everything involved.

## 3b. matplotlib.pyplot: plotting

```python
import matplotlib.pyplot as plt
```

`matplotlib` is the plotting library; `pyplot` is its simple, MATLAB-flavoured front end.
You build a figure by calling functions one after another, each adding to the "current"
plot, then finish with `plt.show()`.

### A complete plot

```python
x = np.linspace(0, 2 * np.pi, 200)

plt.figure(figsize=(8, 4))
plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), "--", label="cos(x)")     # dashed
plt.xlabel("x (radians)")
plt.ylabel("amplitude")
plt.title("Sine and cosine")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

Every one of those lines matters, and the three that students skip are the three that lose
marks: **`xlabel`, `ylabel`, `title`**. A plot without axis labels is not a result, it is a
decoration. Add `legend()` whenever there is more than one series.

### The main plot types

| Function | Use for |
|---|---|
| `plt.plot(x, y)` | lines, trends, functions, time series |
| `plt.scatter(x, y)` | individual points; relationships between two variables |
| `plt.hist(data, bins=20)` | distribution of one variable |
| `plt.bar(labels, heights)` | categorical comparisons |
| `plt.errorbar(x, y, yerr=e)` | measurements with uncertainty |
| `plt.imshow(matrix)` | 2D arrays, heatmaps, images |

### Styling

Format strings combine colour, marker and line style: `"ro-"` is red circles joined by a
line, `"k--"` is a black dashed line, `"g."` is green dots. Or be explicit with keyword
arguments: `color="tab:blue"`, `linewidth=2`, `marker="o"`, `alpha=0.5`, `linestyle=":"`.

### Several plots in one figure

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))     # 1 row, 2 columns
axes[0].plot(x, np.sin(x))
axes[0].set_title("sine")                            # note: set_title, not title
axes[1].hist(data, bins=20)
axes[1].set_title("histogram")
fig.tight_layout()
plt.show()
```

This is the **object-oriented interface**, and it is what you should graduate to: `fig` is
the whole canvas, each `ax` is one set of axes. On an `ax`, the setters are named
`set_xlabel`, `set_ylabel`, `set_title`. `tight_layout()` stops labels overlapping.

### Saving

```python
plt.savefig("figure.png", dpi=300, bbox_inches="tight")
```

Call it **before** `plt.show()`: showing a figure clears it, and you will save a blank
image otherwise. Use `.png` for reports and `.pdf` or `.svg` for anything that must stay
sharp when zoomed.

## 3c. pandas: labelled data tables

```python
import pandas as pd
```

numpy gives you a grid of numbers. **pandas** gives you a grid of numbers with *column
names, row labels, mixed types and missing values*; in other words, a real dataset. If
you have ever used a spreadsheet, pandas is that, driven by code.

Two objects:

- **`Series`**: one labelled column.
- **`DataFrame`**: a whole table: a dict of Series sharing an index.

### Creating a DataFrame

```python
df = pd.DataFrame({
    "name":  ["Ada", "Alan", "Grace", "Katherine"],
    "dept":  ["maths", "maths", "navy", "nasa"],
    "score": [96, 88, 99, 93],
})

df = pd.read_csv("data.csv")        # in practice, almost always this
```

`read_csv` has a sibling for nearly every format: `read_excel`, `read_json`,
`read_parquet`, `read_sql`.

### Looking at it first, always

```python
df.head()          # first 5 rows          <- do this every single time
df.tail(3)
df.shape           # (rows, columns)
df.columns
df.dtypes
df.info()          # types plus non-null counts, in one view
df.describe()      # count/mean/std/min/quartiles/max for numeric columns
```

### Selecting

```python
df["score"]                 # one column -> a Series
df[["name", "score"]]       # several columns -> a DataFrame (note the double brackets)

df.iloc[0]                  # row by POSITION
df.iloc[0:3]                # first three rows
df.loc[2, "name"]           # row by LABEL, and a column by name
```

`loc` is label-based, `iloc` is integer-position-based. Remember them as
**l**abel and **i**nteger.

### Filtering: the workhorse

```python
df[df["score"] > 90]
df[(df["score"] > 90) & (df["dept"] == "maths")]
```

Same boolean-mask idea as numpy. Two rules that trip everyone up: use `&` and `|`, not
`and`/`or`, and **wrap every condition in parentheses**: the operator precedence will
bite you otherwise.

### Adding, sorting, grouping

```python
df["passed"] = df["score"] >= 90              # a new column, computed row-wise
df["scaled"] = df["score"] / df["score"].max()

df.sort_values("score", ascending=False)

df.groupby("dept")["score"].mean()            # split, apply, combine
df.groupby("dept").agg(n=("score", "size"), avg=("score", "mean"))
```

`groupby` is the heart of pandas: split the rows into groups, apply an aggregation to
each, combine the answers into a new table. Almost every "average X per Y" question is
one `groupby`.

### Missing data

Missing values show up as `NaN`. `df.isna().sum()` counts them per column,
`df.dropna()` removes those rows, `df.fillna(0)` (or `.fillna(df.mean())`) replaces them.
Deciding *which* is appropriate is a judgement call about your data, not a technicality.

### Tensors

A **tensor** is numpy's `ndarray` with two extra abilities: it can live on a **GPU**, and
it can **remember how it was computed** so gradients can be worked out automatically.
Everything you learned about numpy carries across:

```python
t = torch.tensor([1.0, 2.0, 3.0])
torch.zeros(2, 3), torch.ones(2, 3), torch.arange(10), torch.linspace(0, 1, 5)
torch.rand(2, 3)              # uniform [0,1)
torch.randn(2, 3)             # standard normal

t.shape, t.dtype, t.device
t * 2, t + t, t @ t           # same vectorised operations
t.mean(), t.sum(), t.max()
t.reshape(3, 1)

torch.from_numpy(arr)         # numpy -> torch (shares memory!)
t.numpy()                     # torch -> numpy
```

Two differences worth knowing early: the default float type is **float32** (numpy uses
float64), and neural network layers insist on float; a `long`/int tensor going into a
model is one of the most common torch errors you will hit.

### Devices

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
t = t.to(device)
```

Both operands of an operation must be on the same device, or you get a runtime error.

### Autograd: the actual magic

Set `requires_grad=True` and torch records every operation applied to a tensor. Call
`.backward()` on the final scalar result and it works backwards through that record,
filling in the derivative with respect to every input.

```python
x = torch.tensor(3.0, requires_grad=True)
y = x ** 2 + 2 * x + 1        # y = 16
y.backward()                  # compute dy/dx
x.grad                        # tensor(8.), and indeed dy/dx = 2x + 2 = 8 at x = 3
```

That is the whole idea behind training a neural network: define how wrong you are (the
**loss**), ask autograd for the gradient of the loss with respect to every parameter, then
nudge each parameter downhill. Repeat a few thousand times.

### The training loop

Every PyTorch training loop, from a two-line regression to a large language model, has the
same five steps:

```python
model = torch.nn.Linear(1, 1)                                  # 1 input -> 1 output
loss_fn = torch.nn.MSELoss()
optimiser = torch.optim.SGD(model.parameters(), lr=0.01)

for epoch in range(200):
    prediction = model(X)                  # 1. forward pass
    loss = loss_fn(prediction, y)          # 2. how wrong are we?
    optimiser.zero_grad()                  # 3. clear last round's gradients
    loss.backward()                        # 4. compute new gradients
    optimiser.step()                       # 5. nudge the parameters
```

Step 3 is not optional: torch *accumulates* gradients by default, so forgetting
`zero_grad()` silently ruins the training. It is the single most common PyTorch bug.

Do not worry about mastering this today. Recognise the shape of it; you will see this
same loop again all week.

| Symptom | Cause |
|---|---|
| `ModuleNotFoundError` after a successful install | you installed into a different environment; use `%pip`, and check `sys.executable` |
