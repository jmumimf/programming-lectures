# 1. Doing math

The simplest way to think about Python is as a very good calculator that never gets tired.
You type an expression, Python evaluates it, and hands you back a value.

## The arithmetic operators

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | addition | `7 + 3` | `10` |
| `-` | subtraction | `7 - 3` | `4` |
| `*` | multiplication | `7 * 3` | `21` |
| `/` | division (always a decimal) | `7 / 3` | `2.3333333333333335` |
| `//` | floor division (throws away the remainder) | `7 // 3` | `2` |
| `%` | modulo (*only* the remainder) | `7 % 3` | `1` |
| `**` | exponent / power | `7 ** 3` | `343` |

Two of these surprise everybody at first:

- **`/` always gives a decimal**, even when it divides evenly. `6 / 3` is `2.0`, not `2`.
- **`%` (modulo) gives the leftover.** `17 % 5` is `2`, because 17 is three fives with 2
  left over. This is the standard trick for "is this number even?" (`n % 2 == 0`) and for
  wrapping a value around a clock.

## Order of operations

Python follows normal maths precedence: **P**arentheses, **E**xponents,
**M**ultiplication/**D**ivision, **A**ddition/**S**ubtraction:

```python
2 + 3 * 4        # 14, not 20, the * happens first
(2 + 3) * 4      # 20, parentheses win
2 ** 3 ** 2      # 512, because ** groups right-to-left: 2 ** (3 ** 2)
-3 ** 2          # -9, because ** binds tighter than the minus sign
```

When in doubt, **add parentheses**. They are free, and they make your intent obvious to
the next person reading the code (usually you, three weeks from now).

### A wart worth knowing about: floating point

Computers store decimals in binary, and some perfectly ordinary decimals (like `0.1`)
have no exact binary representation; the same way `1/3` has no exact decimal one. So
arithmetic on decimals is very slightly approximate.

This is **not a Python bug**; essentially every language on every CPU behaves this way.
The practical rule: **never compare decimals with `==`.** Compare them with a tolerance
instead (`abs(a - b) < 1e-9`). You will meet this idea again all week in numerical methods.

# 2. Variables and types

## Variables are names for values

A variable is a **name pointing at a value**. You create one with `=`, which is not
"equals" in the mathematical sense; it means *"evaluate the thing on the right, and give
it this name."*

```python
radius = 5
area = 3.14159 * radius ** 2
radius = 10        # the name now points at a new value; `area` does NOT update
```

That last line matters. `area` was computed **once**, from the old `radius`. Python does
not remember the formula, and nothing recalculates unless you run the line again. This is
the single most common source of confusion when you start.

## Naming rules and conventions

**Rules** (break these and you get a `SyntaxError`):

- letters, digits and underscores only; cannot start with a digit
- case-sensitive; `count` and `Count` are two different variables
- cannot be a reserved keyword (`if`, `for`, `class`, `return`, `None`, `import`, ...)

**Conventions** (break these and nothing crashes, but people judge you):

- `snake_case` for variables and functions: `student_count`, not `studentCount`
- `ALL_CAPS` for constants: `MAX_ITERATIONS = 1000`
- names should say what the thing *is*. `n` and `tmp` are fine for two lines; `velocity`
  is much better for twenty.

Also avoid shadowing built-ins: naming a variable `list`, `sum`, `str`, `max` or `type`
quietly breaks those functions for the rest of your session.

## The core built-in types

| Type | What it is | Examples |
|---|---|---|
| `int` | whole number, unlimited size | `0`, `-7`, `10**100` |
| `float` | decimal number | `3.14`, `-0.5`, `2.0`, `1e-9` |
| `str` | text (a "string" of characters) | `"hello"`, `'world'`, `""` |
| `bool` | truth value | `True`, `False` (capitalised!) |
| `NoneType` | "no value at all" | `None` |

Python is **dynamically typed**: you never declare a type, and a name can point at a
different type later. Ask what something is with `type(x)`.

## Converting between types

`int("42")`, `float("3.5")`, `str(42)`, `bool(0)`. These conversions can fail loudly:
`int("hello")` raises a `ValueError`. And note that `int(3.9)` gives `3`: it **truncates
toward zero**, it does not round. (Use `round(3.9)` if you want 4.)

## Strings, briefly

Quotes can be single or double; pick one and be consistent. A few essentials:

```python
name = "Ada"
len(name)                 # 3
name + " Lovelace"        # concatenation -> "Ada Lovelace"
"ha" * 3                  # repetition    -> "hahaha"
name.upper()              # "ADA"
"  padded  ".strip()      # "padded"
"a,b,c".split(",")        # ["a", "b", "c"]
"Ada" in "Ada Lovelace"   # True
```

Strings are **immutable**: every string method returns a *new* string and leaves the
original untouched. Calling `name.upper()` on its own line does nothing lasting; you
have to assign the result somewhere.

## Augmented assignment

`count = count + 1` is so common that it has a shorthand: `count += 1`.
The same works for `-=`, `*=`, `/=`, `//=`, `%=` and `**=`.

# 3. Printing, the terminal, and comments

## `print()`

A notebook shows you the last expression of a cell for free. Everywhere else (and any
time you want to see **more than one** thing), you need `print()`.

```python
print("hello")
print("x is", 5, "and y is", 7)      # multiple arguments, separated by spaces
print("a", "b", "c", sep="-")        # a-b-c
print("no newline after me", end="")
```

`print` **displays** a value; it does not **produce** one. That distinction becomes
critical in section 6, where we compare `print` with `return`.

## f-strings, the good way to build text

Put an `f` before the opening quote and you can drop expressions straight into `{}`:

```python
name, score = "Ada", 96.5432
print(f"{name} scored {score}")            # Ada scored 96.5432
print(f"{name} scored {score:.1f}")        # Ada scored 96.5
print(f"{score:.2f}% over {2 * 3} tests")  # any expression works inside the braces
```

Useful format specifiers, written after a colon:

| Spec | Meaning | Example output |
|---|---|---|
| `:.2f` | 2 decimal places | `3.14` |
| `:8.2f` | width 8, right-aligned | `    3.14` |
| `:<8` `:>8` `:^8` | left / right / centre within width 8 | |
| `:,` | thousands separators | `1,048,576` |
| `:e` | scientific notation | `1.05e+06` |
| `:.0%` | as a percentage | `42%` |

f-strings are how you build readable output tables, which you will want constantly.

## Getting input

`input("prompt: ")` shows a prompt and waits for the user to type something.
**It always returns a string**: even if they type `42`, you get `"42"`. So convert it:
`age = int(input("Age: "))`.

In a notebook, `input()` pops a box up at the top of the window and blocks the cell until
you answer, which is mildly annoying. For practice you can hardcode a value instead.

## Comments

`#` starts a comment and Python ignores the rest of the line. Comments should explain
**why**, not **what**; the code already says what.

```python
# BAD:  add one to i
i += 1

# GOOD: skip the header row of the file
i += 1
```

## Scripts and the terminal

A notebook is not the only way to run Python. Put code in a file called `hello.py` and
run it from a terminal with:

```
python hello.py
```

In a script nothing, is displayed automatically; `print()` is your only window into the
program. Notebooks vs `.py` files is the first topic of the **ecosystem** notebook.

# 4. Making decisions: `if` / `elif` / `else`

So far every program has run straight from top to bottom. Conditionals let it choose.

## Comparison operators

These produce a `bool` (`True` or `False`):

| Operator | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal to |
| `<` `<=` | less than, less than or equal |
| `>` `>=` | greater than, greater than or equal |
| `in` | is contained in (`"a" in "cat"`, `3 in [1, 2, 3]`) |

> **`=` vs `==` is the classic beginner bug.** One `=` *assigns*, two `==` *asks a question*.

Python also lets you chain comparisons the way maths does: `0 <= score <= 100` means
exactly what it looks like.

## Logical operators

`and`, `or`, `not`; words, not symbols:

```python
is_weekend and is_sunny          # True only if BOTH are True
temperature < 0 or temperature > 40
not raining
```

## The `if` statement

```python
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"
```

Things to notice:

- the condition is followed by a **colon**
- the body is **indented** 4 spaces, and in Python indentation is *syntax*, not a stylistic choice.
  Get it wrong and you get an `IndentationError` (or, worse, working code that does the
  wrong thing).
- `elif` means "else if". You can have as many as you like, or none.
- `else` is optional and takes no condition.
- **Order matters.** Python checks the branches top to bottom and runs the *first* one
  that is `True`, then skips the rest. If you put `score >= 70` first, everybody gets a C.

## Truthiness

Any value can be used as a condition. These are all "falsy":

`False`, `None`, `0`, `0.0`, `""`, `[]`, `{}`, `()`

Everything else is "truthy". So `if items:` is the idiomatic way to say
"if the list is not empty".

# 5. Loops

A loop repeats a block of code. Python has two kinds, and choosing between them is easy:

- **`for`**: when you know *what you are looping over* (a range of numbers, the items in
  a list). This is the one you will use 90% of the time.
- **`while`**: when you loop *until some condition changes* and you cannot say in advance
  how many repetitions that will take.

## `for` and `range`

```python
for i in range(5):
    print(i)          # starts at 0, STOPS BEFORE 5
```

`range` comes in three forms:

| Call | Produces |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(0, 10, 3)` | 0, 3, 6, 9 |
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 |

The stop value is always **excluded**. This is the source of the famous *off-by-one*
error: to loop over 1 to 100 inclusive you need `range(1, 101)`.

`for` also walks straight over any collection; you will use this constantly in section 7:

```python
for letter in "cat":
    print(letter)
```

## `while`

```python
count = 3
while count > 0:
    print(count)
    count -= 1        # <-- if you forget this line, the loop NEVER ends
print("liftoff")
```

The condition is checked *before* each pass. If it is never made false, you get an
**infinite loop**; stop it with the ■ (interrupt) button in the toolbar, or
Ctrl+C in a terminal. Everyone does this at least once.

## The accumulator pattern

The single most useful loop shape there is: start with an empty/zero value, then update it
once per pass.

```python
total = 0
for i in range(1, 11):
    total += i
print(total)          # 55
```

## `break` and `continue`

- `break` leaves the loop immediately.
- `continue` skips the rest of *this* pass and goes to the next one.

## Nested loops

A loop inside a loop. The inner loop runs completely for *every single* pass of the outer
one, so two nested loops over 1000 items each is a million iterations. Be aware of that.

```python
for row in range(3):
    for col in range(3):
        print(row, col)
```

# 6. Functions

A function is a named, reusable piece of code. You write it once and call it as often as
you like. Functions exist for three reasons: to avoid repeating yourself, to give a
complicated operation a name, and to let you test one small piece in isolation.

## Defining and calling

```python
def greet(name):
    return f"Hello, {name}!"

message = greet("Ada")        # call it
print(message)                # Hello, Ada!
```

- `def` starts the definition; the body is indented, like everything else in Python.
- `name` is a **parameter** (the placeholder). `"Ada"` is an **argument** (the real value).
- **Defining a function runs none of its code.** Nothing happens until you *call* it.

## `return` vs `print`: the distinction that trips everyone

`print` shows a value to a human. `return` hands a value back to the *program* so you can
keep computing with it.

```python
def double_bad(x):
    print(x * 2)              # displays it, then throws it away

def double_good(x):
    return x * 2

y = double_good(5) + 1        # 11 (works)
z = double_bad(5) + 1         # TypeError: double_bad returned None
```

A function with no `return` returns `None`. `return` also exits the function immediately;
any code after it in the same branch never runs.

## Default and keyword arguments

```python
def power(base, exponent=2):        # exponent defaults to 2
    return base ** exponent

power(5)                            # 25
power(5, 3)                         # 125
power(exponent=3, base=5)           # 125; named arguments, order stops mattering
```

Parameters with defaults must come after those without.

## Returning several values

```python
def min_and_max(a, b, c):
    return min(a, b, c), max(a, b, c)

low, high = min_and_max(3, 9, 5)    # 3, 9
```

Technically it returns one tuple, which you then unpack; see section 7.

## Scope

Variables created inside a function are **local**: they exist only while the function
runs, and are invisible outside it. A function *can* read variables from outside, but
relying on that makes code hard to follow. Prefer passing values in as parameters and
handing results back with `return`.

## Docstrings and type hints

A string on the first line of a function is its **docstring**; what `help(f)` shows.
Type hints (`x: float -> float`) are optional annotations that document intent; Python
does not enforce them, but editors use them to catch mistakes.

```python
def kinetic_energy(mass: float, velocity: float) -> float:
    """Return the kinetic energy of a body in joules."""
    return 0.5 * mass * velocity ** 2
```

# 7. Data storage types

One variable holds one value. Real programs need to hold *collections* of values. Python
gives you four workhorse containers, and picking the right one is most of the battle.

| Type | Written as | Ordered? | Changeable? | Use it when... |
|---|---|---|---|---|
| `list` | `[1, 2, 3]` | yes | yes | you have a sequence of things that may grow or change |
| `tuple` | `(1, 2, 3)` | yes | **no** | the group is fixed; a coordinate, a returned pair |
| `dict` | `{"a": 1}` | yes (insertion) | yes | you look things up **by name/key** |
| `set` | `{1, 2, 3}` | no | yes | you only care about membership and uniqueness |

## 7a. Lists

```python
scores = [88, 92, 79, 95]
```

### Indexing and slicing

Indices start at **0**. Negative indices count from the end.

```python
scores[0]      # 88   first
scores[-1]     # 95   last
scores[1:3]    # [92, 79]   from 1 up to but NOT including 3
scores[:2]     # [88, 92]   from the start
scores[2:]     # [79, 95]   to the end
scores[::-1]   # [95, 79, 92, 88]   reversed copy
```

Slicing follows the same "stop is excluded" rule as `range`, and always gives you a
**new** list. An index that does not exist raises an `IndexError`.

### Lists are mutable

```python
scores[0] = 100          # change in place
scores.append(70)        # add one item to the end
scores.extend([60, 65])  # add several
scores.insert(0, 99)     # insert at a position
scores.pop()             # remove AND return the last item
scores.remove(79)        # remove the first item equal to 79
scores.sort()            # sorts IN PLACE, returns None
sorted(scores)           # returns a NEW sorted list, original untouched
len(scores)              # how many items
sum(scores), min(scores), max(scores)
79 in scores             # membership test -> bool
```

> **Gotcha:** `scores = scores.sort()` sets `scores` to `None`, because `.sort()` sorts in
> place and returns nothing. Use `sorted()` if you want a value back.

### Looping over lists

```python
for score in scores:              # the items themselves
    print(score)

for i, score in enumerate(scores):     # index AND item
    print(i, score)

for name, score in zip(names, scores): # two lists in lockstep
    print(name, score)
```

Prefer `for item in items:` over `for i in range(len(items)):`. Use `enumerate` only when
you genuinely need the position.

### The aliasing gotcha

```python
a = [1, 2, 3]
b = a            # NOT a copy: both names point at the SAME list
b.append(4)
print(a)         # [1, 2, 3, 4]   <- surprise!

c = a.copy()     # this IS a copy (a[:] and list(a) also work)
```

## 7b. Tuples

A tuple is an immutable list, written with parentheses (or with nothing at all):

```python
point = (3, 4)
point = 3, 4          # the parentheses are optional
point[0]              # 3, indexing works the same
point[0] = 9          # TypeError: tuples cannot be changed
```

Use a tuple when the group of values is fixed and its *shape* means something: an (x, y)
coordinate, an (r, g, b) colour, a row from a table. The immutability is a feature: it
guarantees nobody can modify it by accident, and it lets tuples be used as dictionary keys.

### Unpacking

This is where tuples really earn their place:

```python
x, y = point                     # x = 3, y = 4
a, b = b, a                      # the swap trick from section 2: it is a tuple!
low, high = min_and_max(...)     # multiple return values from section 6
for i, item in enumerate(items): # enumerate yields tuples
```

A one-element tuple needs a trailing comma: `(5,)`. Just `(5)` is the number 5.

## 7c. Dictionaries

A dictionary stores **key → value** pairs. Instead of asking "what is at position 3?" you
ask "what is stored under `'Ada'`?"; which is far more often the question you actually
have.

```python
ages = {"Ada": 36, "Alan": 41, "Grace": 45}

ages["Ada"]                # 36
ages["Ada"] = 37           # update
ages["Katherine"] = 101    # add a new pair; same syntax
del ages["Alan"]           # remove
len(ages)                  # 3
"Grace" in ages            # True; checks KEYS
```

Asking for a key that is not there raises a `KeyError`. When that is a real possibility,
use `.get()`, which returns `None` (or a default you choose) instead of exploding:

```python
ages.get("Nobody")          # None
ages.get("Nobody", 0)       # 0
```

### Looping over a dictionary

```python
for name in ages:                    # keys (the default)
    print(name)

for name, age in ages.items():       # keys AND values; the usual one
    print(f"{name} is {age}")

ages.keys(), ages.values(), ages.items()
```

Keys must be immutable (strings, numbers, tuples; not lists) and unique. Values can be
anything at all, **including other lists and dicts**, which is how you build up structured
data:

```python
gradebook = {
    "Ada":  {"quiz": [9, 10], "exam": 96},
    "Alan": {"quiz": [8,  7], "exam": 88},
}
gradebook["Ada"]["quiz"][1]      # 10
```

### The counting pattern

Worth memorising; you will write it constantly:

```python
counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1
```

## 7d. Sets

A set is an **unordered** collection of **unique** items:

```python
s = {3, 1, 4, 1, 5, 9, 2, 6, 5}     # {1, 2, 3, 4, 5, 6, 9}; duplicates vanish
set([1, 1, 2, 3])                   # the standard way to de-duplicate a list
```

Two reasons to reach for a set:

1. **Deduplication**; `list(set(items))` removes repeats (but loses the order).
2. **Fast membership tests**; `x in some_set` is effectively instant no matter how big
   the set is, whereas `x in some_list` has to scan every element.

Set maths works with operators or methods:

```python
a, b = {1, 2, 3}, {3, 4, 5}
a | b      # union         {1, 2, 3, 4, 5}
a & b      # intersection  {3}
a - b      # difference    {1, 2}
a ^ b      # symmetric difference {1, 2, 4, 5}
```

Note that `{}` is an empty **dict**, not an empty set; you need `set()` for that.

## 7e. Comprehensions

A comprehension is a compact way to build a list (or dict, or set) from another
collection. These two blocks do exactly the same thing:

```python
# the long way
squares = []
for n in range(10):
    squares.append(n ** 2)

# the comprehension
squares = [n ** 2 for n in range(10)]
```

You can filter with a trailing `if`:

```python
evens = [n for n in range(20) if n % 2 == 0]
names_upper = [name.upper() for name in names if len(name) > 3]
```

Dict and set comprehensions use the same shape:

```python
{name: len(name) for name in names}
{word[0] for word in words}
```

Comprehensions are idiomatic Python, and you will see them everywhere. But readability
wins: if you find yourself nesting two of them with a filter, write the plain loop instead.

## Appendix: errors you will definitely meet

| Error | Usually means |
|---|---|
| `SyntaxError` | a typo; a missing `:`, an unclosed bracket or quote |
| `IndentationError` | your indentation is inconsistent (mixed tabs and spaces, or a missing indent after `:`) |
| `NameError` | you used a name that does not exist yet; a typo, or you never ran the cell that defined it |
| `TypeError` | wrong kind of thing; e.g. `"a" + 1`, or calling something that is not a function |
| `ValueError` | right type, impossible value; e.g. `int("hello")` |
| `IndexError` | list index out of range (remember: last valid index is `len(x) - 1`) |
| `KeyError` | that key is not in the dictionary; consider `.get()` |
| `ZeroDivisionError` | you divided by zero |
| `AttributeError` | that type has no such method; e.g. `.append()` on a string |

**How to read a traceback:** start at the **bottom**. The last line names the error and
explains it; the lines above show the path of calls that got you there. The arrow `--->`
points at the offending line.
